## Operações com Vetores - Adição e Subtração

```python
# Definindo dois vetores como listas
vetor_a = [3, 4, 5]
vetor_b = [1, 2, 3]
```

```python
# Função para adição de vetores
def adicionar_vetores(a, b):
    return [a[i] + b[i] for i in range(len(a))]
```

```python
# Função para subtração de vetores
def subtrair_vetores(a, b):
    return [a[i] - b[i] for i in range(len(a))]
```

```python
# Realizando as operações
vetor_soma = adicionar_vetores(vetor_a, vetor_b)
vetor_diferenca = subtrair_vetores(vetor_a, vetor_b)
```

```python
# Mostrando os resultados
print("Vetor A:", vetor_a)
print("Vetor B:", vetor_b)
print("Soma dos Vetores:", vetor_soma)
print("Diferença dos Vetores:", vetor_diferenca)
```

## Operações com Vetores - Cross Product (Produto Vetorial)

O produto vetorial entre dois vetores resulta em um terceiro vetor que é perpendicular aos dois vetores originais.

```python
# Função para calcular o produto vetorial entre 2 vetores
def produto_vetorial(v1, v2):
    x = v1[1] * v2[2] - v1[2] * v2[1]
    y = v1[2] * v2[0] - v1[0] * v2[2]
    z = v1[0] * v2[1] - v1[1] * v2[0]
    return [x, y, z]
```

```python
# Cria os vetores
vetor_a = [2, -3, 1]
vetor_b = [5, 7, 6]
```

```python
# Calculando o produto vetorial
resultado = produto_vetorial(vetor_a, vetor_b)
print("Produto Vetorial:", resultado)
```

## Operações com Vetores - Dot Product (Produto Escalar)

O produto escalar de dois vetores é uma operação que resulta em um escalar (um número). Matematicamente, é a soma dos produtos dos componentes correspondentes dos vetores.

```python
# Função para calcular o produto escalar
def produto_escalar(v1, v2):
    return sum(a * b for a, b in zip(v1, v2))
```

```python
# Cria os vetores
vetor_a = [2, -3, 1]
vetor_b = [5, 7, 6]
```

```python
# Calculando o produto escalar
resultado = produto_escalar(vetor_a, vetor_b)
print("Produto Escalar:", resultado)
```

### Função de Custo MSE com Operações Entre Vetores

Esta função dsa_func_custo é uma implementação da função de custo de erro quadrático médio, que é amplamente usada em regressão linear. O objetivo dessa função é medir quão bem o modelo de regressão (representado pelos parâmetros theta) está desempenhando em relação ao conjunto de dados (X e y). Um custo mais baixo indica que o modelo está fazendo previsões mais precisas.

```python
# Função de Custo MSE com Operações Entre Vetores
def dsa_func_custo(X, y, theta):

    # m é o número de exemplos de treinamento (observações) no conjunto de dados
    m = len(y)

    # Calcula a função de custo para a regressão linear
    # A função de custo é a média dos erros quadrados entre as previsões e os valores reais
    # (1/(2*m)) é um fator de escala comum na função de custo do erro quadrático médio
    # np.sum(np.square(X.dot(theta) - y)) é a soma dos quadrados dos erros
    # X.dot(theta) é o produto escalar de X (dados de entrada) e theta (parâmetros do modelo)
    # X.dot(theta) - y é a diferença entre as previsões e os valores reais
    cost = (1/(2*m)) * np.sum(np.square(X.dot(theta) - y))

    # Retorna o custo calculado
    return cost
```

### Algoritmo Gradiente Descendente com Operações Entre Vetores

```python
# Gradiente descendente
def dsa_gradient_descent(X, y, theta, learning_rate, iterations):

    # m é o número de exemplos de treinamento no conjunto de dados
    m = len(y)

    # Inicializa um array para armazenar o histórico de custo em cada iteração
    cost_history = np.zeros(iterations)

    # Loop para realizar o número especificado de iterações
    for i in range(iterations):

        # Calcula o gradiente da função de custo em relação a theta
        # O gradiente é a média dos produtos dos erros e as características de entrada
        gradients = 1/m * X.T.dot(X.dot(theta) - y)

        # Atualiza os parâmetros theta subtraindo o produto da taxa de aprendizado e o gradiente
        # Esta é a etapa de descida do gradiente, movendo-se em direção ao mínimo da função de custo
        theta = theta - learning_rate * gradients

        # Calcula o custo com o theta atualizado e armazena no histórico de custo
        cost_history[i] = dsa_func_custo(X, y, theta)

    # Retorna os parâmetros theta otimizados e o histórico de custo
    return theta, cost_history
```

### Treinamento do Modelo

```python
# Parâmetros iniciais

# Inicializa os parâmetros randomicamente
theta = np.random.randn(2, 1)

# Taxa de aprendizado
learning_rate = 0.01

# Número de passadas de treino
iterations = 1000
```

```python
# Treinando o modelo
theta_final, cost_history = dsa_gradient_descent(X_b, y, theta, learning_rate, iterations)
```

### Exercício 3: Norma de um Vetor
Descrição: Escreva uma função que calcule a norma (ou comprimento) de um vet

### Explicação da Norma de um Vetor

A **norma de um vetor** (também conhecida como magnitude ou comprimento) é uma medida do seu "tamanho" no espaço vetorial. Para um vetor em um espaço Euclidiano, a norma mais comum é a **norma Euclidiana** (ou L2-norma).

Matematicamente, para um vetor \( \mathbf{v} = [v_1, v_2, ..., v_n] \), a norma Euclidiana é calculada como a raiz quadrada da soma dos quadrados de seus componentes:

$$ ||\mathbf{v}|| = \sqrt{v_1^2 + v_2^2 + ... + v_n^2} $$

No contexto geométrico, se você pensar em um vetor como uma seta de um ponto de origem até outro, a norma é o comprimento dessa seta. É sempre um valor não negativo e é zero apenas se o vetor for o vetor nulo (todos os componentes são zero).

Para o Exercício 3, vamos implementar essa norma Euclidiana.

```python
# Importa a função sqrt (raiz quadrada) do módulo math
import math

def norma_vetor(v):
  # Calcula a soma dos quadrados de cada componente do vetor
  soma_dos_quadrados = sum([x**2 for x in v])

  # A norma é a raiz quadrada dessa soma
  return math.sqrt(soma_dos_quadrados)

# --- Demonstração da função ---

# Exemplo 1: Um vetor simples
vetor1 = [3, 4] # A norma esperada é sqrt(3^2 + 4^2) = sqrt(9 + 16) = sqrt(25) = 5
print(f"Vetor 1: {vetor1}")
print(f"Norma do Vetor 1: {norma_vetor(vetor1)}\n")

# Exemplo 2: Usando um dos vetores anteriores (vetor_a = [2, -3, 1])
vetor2 = [2, -3, 1]
print(f"Vetor 2: {vetor2}")
print(f"Norma do Vetor 2: {norma_vetor(vetor2)}\n")

# Exemplo 3: Outro vetor
vetor3 = [1, 2, 2]
print(f"Vetor 3: {vetor3}")
print(f"Norma do Vetor 3: {norma_vetor(vetor3)}")
```

### Exercício 4: Ângulo entre Dois Vetores
Descrição: Implemente uma função que calcule o ângulo (em graus) entre dois vetore

### Explicação do Ângulo entre Dois Vetores

O **ângulo entre dois vetores** é uma medida da sua orientação relativa no espaço. Ele é fundamental em diversas áreas, como física, engenharia e ciência de dados.

Para dois vetores \( \mathbf{v_1} \) e \( \mathbf{v_2} \), o cosseno do ângulo \( \theta \) entre eles pode ser calculado usando a seguinte fórmula, que envolve o **produto escalar** e as **normas** dos vetores:

$$ \cos(\theta) = \frac{\mathbf{v_1} \cdot \mathbf{v_2}}{||\mathbf{v_1}|| \cdot ||\mathbf{v_2}||} $$

Onde:
*   \( \mathbf{v_1} \cdot \mathbf{v_2} \) é o produto escalar dos vetores \( \mathbf{v_1} \) e \( \mathbf{v_2} \).
*   \( ||\mathbf{v_1}|| \) é a norma (ou magnitude) do vetor \( \mathbf{v_1} \).
*   \( ||\mathbf{v_2}|| \) é a norma (ou magnitude) do vetor \( \mathbf{v_2} \).

Para encontrar o ângulo \( \theta \) em si, aplicamos a função arco cosseno (arccos) ao resultado:

$$ \theta = \arccos\left(\frac{\mathbf{v_1} \cdot \mathbf{v_2}}{||\mathbf{v_1}|| \cdot ||\mathbf{v_2}||}\right) $$

---

### Entendendo o Arco Cosseno (arccos ou cos⁻¹)

O **arco cosseno** (frequentemente abreviado como `arccos` ou `cos^-1`) é a função inversa do cosseno.

Em termos simples:

*   Se você tem um **ângulo** e quer encontrar o **cosseno** desse ângulo, você usa a função `cosseno` (por exemplo, `math.cos(0)` resulta em 1).
*   Se você tem um **valor de cosseno** (que estará sempre entre -1 e 1) e quer descobrir qual **ângulo** gerou aquele cosseno, você usa a função **arco cosseno**.

Por exemplo:

*   Sabemos que o cosseno de 0 graus (ou 0 radianos) é 1. Portanto, o arco cosseno de 1 é 0 graus.
*   Sabemos que o cosseno de 90 graus (ou π/2 radianos) é 0. Portanto, o arco cosseno de 0 é 90 graus.
*   Sabemos que o cosseno de 180 graus (ou π radianos) é -1. Portanto, o arco cosseno de -1 é 180 graus.

Em Python, a função `math.acos()` retorna o ângulo em **radianos**. Para convertê-lo para graus, usamos `math.degrees()`.

---

O resultado do arco cosseno é geralmente em radianos, então precisaremos convertê-lo para graus, conforme solicitado pelo exercício. O código Python usará as funções `produto_escalar` e `norma_vetor` que já definimos anteriormente.

```python
import math

def angulo_entre_vetores(v1, v2):
  # Certifica-se de que os vetores têm o mesmo comprimento
  if len(v1) != len(v2):
    raise ValueError("Os vetores devem ter o mesmo número de dimensões.")

  # Calcula o produto escalar (reutilizando a função já definida)
  # A função produto_escalar já foi definida anteriormente no notebook.
  # Se a célula não estiver visível, ela está mais acima.
  produto_escalar_val = produto_escalar(v1, v2)

  # Calcula a norma de cada vetor (reutilizando a função já definida)
  # A função norma_vetor já foi definida anteriormente no notebook.
  # Se a célula não estiver visível, ela está mais acima.
  norma_v1 = norma_vetor(v1)
  norma_v2 = norma_vetor(v2)

  # Verifica se alguma norma é zero para evitar divisão por zero
  if norma_v1 == 0 or norma_v2 == 0:
    return 0.0 # Vetor nulo não tem direção, então o ângulo é considerado 0 ou indefinido

  # Calcula o cosseno do ângulo
  cos_theta = produto_escalar_val / (norma_v1 * norma_v2)

  # Garante que o valor de cos_theta esteja no intervalo [-1, 1] devido a imprecisões de ponto flutuante
  cos_theta = max(-1.0, min(1.0, cos_theta))

  # Calcula o ângulo em radianos e depois converte para graus
  angulo_radianos = math.acos(cos_theta)
  angulo_graus = math.degrees(angulo_radianos)

  return angulo_graus

# --- Demonstração da função ---

# Exemplo 1: Vetores paralelos (ângulo 0 graus)
vetor_a = [1, 0]
vetor_b = [2, 0]
print(f"Ângulo entre {vetor_a} e {vetor_b}: {angulo_entre_vetores(vetor_a, vetor_b):.2f} graus")

# Exemplo 2: Vetores perpendiculares (ângulo 90 graus)
vetor_c = [1, 0]
vetor_d = [0, 1]
print(f"Ângulo entre {vetor_c} e {vetor_d}: {angulo_entre_vetores(vetor_c, vetor_d):.2f} graus")

# Exemplo 3: Vetores opostos (ângulo 180 graus)
vetor_e = [1, 0]
vetor_f = [-1, 0]
print(f"Ângulo entre {vetor_e} e {vetor_f}: {angulo_entre_vetores(vetor_e, vetor_f):.2f} graus")

# Exemplo 4: Vetores do notebook anterior para Dot Product
vetor_g = [2, -3, 1]
vetor_h = [5, 7, 6]
print(f"Ângulo entre {vetor_g} e {vetor_h}: {angulo_entre_vetores(vetor_g, vetor_h):.2f} graus")
```
